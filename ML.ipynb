```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
hours = np.array([1.5, 2.0, 2.5, 3.2, 4.0, 5.0, 5.5, 6.5, 7.2, 8.0])
scores = np.array([22.0, 28.0, 31.0, 40.0, 48.0, 56.0, 65.0, 72.0, 78.0, 88.0])
```

```python
def run_linear_regression(m, b, run_number):
    # Calculate predictions
    predictions = m * hours + b

    # Calculate Mean Squared Error
    mse = np.mean((scores - predictions) ** 2)

    # Print results
    print("Run " + str(run_number) + " Results:")
    print("Slope (m): " + str(m))
    print("Intercept (b): " + str(b))
    print("Mean Squared Error: " + str(round(mse, 2)))
    print("-" * 30)

    # Create plot
    plt.figure(figsize=(8, 5))

    # Plot actual data points
    plt.scatter(
        hours,
        scores,
        color="blue",
        label="Actual Student Scores"
    )

    # Create regression line
    x_line = np.linspace(0, 10, 100)
    y_line = m * x_line + b

    plt.plot(
        x_line,
        y_line,
        color="black",
        label="Fit Line: y = " + str(m) + "x + " + str(b)
    )

    # Plot error lines
    for x_val, y_val, pred_val in zip(hours, scores, predictions):
        plt.plot(
            [x_val, x_val],
            [y_val, pred_val],
            color="red",
            linestyle="--"
        )

    # Graph labels and title
    plt.title(
        "Run " + str(run_number) +
        "\nError (MSE): " + str(round(mse, 2))
    )

    plt.xlabel("Study Hours")
    plt.ylabel("Exam Score")

    plt.xlim(0, 10)
    plt.ylim(0, 105)

    plt.legend()
    plt.grid(True)

    plt.show()
```

```python
# Example: Run 1 using sample initial values
run_linear_regression(m=3.0, b=2.0, run_number=1)
```
