```python
n=6
y=[2,4,5,7,9,12]
m=0
x=[0.5,1.5,2.5,3.5,4.5,5.5]
c=0
```

```python
def square(x):
    return x**2
```

```python
def sumation(mylist):
    total = 0
    for i in mylist:
        total += i
    return total
```

```python
def sumationofmul(num1,num2):
    total = 0
    for i in range(len(num1)):
        total += (num1[i] * num2[i])
    return total
```

```python
def sumationofsq(number):
    total = 0
    for i in number:
        total += square(i)
    return total
```

```python
m=(n*sumationofmul(x,y)-sumation(x)*sumation(y))/(n*sumationofsq(x)-square(sumation(x)))
c=(sumation(y)-m*sumation(x))/n
```

```python
print(f"y = {m:.2f}x + {c:.2f}, \n\nwhen x=4.5 y={m*4+c:.3f}")
```
