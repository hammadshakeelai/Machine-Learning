```python
X1 = [10, 2, 5, 12]
X2 = [52, 73, 55, 49]
Y = [1, 0, 0, 1]
```

```python
def calc_z(w1, x1, w2, x2, b): # 𝒛 = 𝒘𝟏𝒙𝟏 + 𝒘𝟐𝒙𝟐 + 𝒃
    return w1 * x1 + w2 * x2 + b # 𝒘𝟏𝒙𝟏 + 𝒘𝟐𝒙𝟐 + 𝒃
```

```python
def sigmoid(z): # 𝜎(𝒛) = 𝒈(𝒛)
    return 1 / (1 + math.exp(-z)) # 𝒈(𝒛) = 𝟏 /𝟏 + 𝒆^−𝒛
```

```python
def g(w1, x1, w2, x2, b):
    return sigmoid(calc_z(w1, x1, w2, x2, b))
```

```python
epochs = 20
w1 = 0
w2 = 0
b= 0
alpha = 0.0001
cost_function_of_w1 = 0
cost_function_of_w2 = 0
cost_function_of_b = 0
jwb = 0
m = len(X1) # number of training examples
for i in range(epochs):
    sumation = 0
    loss_function_wrt_w2 = 0
    loss_function_wrt_w1 = 0
    loss_function_wrt_b = 0
    for j in range(m):
        x1 = X1[j]
        x2 = X2[j]
        y = Y[j]
        # 𝑓𝑤𝑏 𝑥 = 𝒈 𝒛 = 𝒈(𝒘𝟏𝒙𝟏 + 𝒘𝟐𝒙𝟐 + 𝒃)

        loss_function_wrt_w1 += (g(w1, x1, w2, x2, b) - y) * x1
        loss_function_wrt_w2 += (g(w1, x1, w2, x2, b) - y) * x2
        loss_function_wrt_b += g(w1, x1, w2, x2, b) - y
        # --
        sumation += (y * math.log(g(w1, x1, w2, x2, b)))+((1-y) * math.log(1-g(w1, x1, w2, x2, b)))
    # 𝑱(𝒘𝒃) = -𝟏/𝒎 * SUM{(1-m)[(𝒚𝒊)𝒍𝒐𝒈(𝒇𝒘𝒃 (𝒙𝒊))+(𝟏 − 𝒚𝒊)𝒍𝒐𝒈(𝟏 − 𝒇𝒘𝒃 (𝒙𝒊))]
    jwb = -1 / m * sumation 

    print("Epoch: ",i+1," Cost: ",jwb," w1: ",w1," w2: ",w2," b: ",b)
    
    w1 = w1 - alpha * ( (1 / m) * loss_function_wrt_w1)
    w2 = w2 - alpha * ( (1 / m) * loss_function_wrt_w2)
    b = b - alpha * ( (1 / m) * loss_function_wrt_b)
```
