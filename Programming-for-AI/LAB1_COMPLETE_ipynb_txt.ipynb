## 25. Mini Project: Dataset Health Check

Write a reusable function that can inspect any small tabular dataset.

````python
def inspect_dataset(df):
    print('DATASET HEALTH CHECK')
    print('Rows:', df.shape[0])
    print('Columns:', df.shape[1])
    print('Column names:', list(df.columns))
    print('\nMissing values:')
    print(df.isnull().sum())
    print('\nDuplicate rows:')
    print(df.duplicated().sum())
    print('\nData types:')
    print(df.dtypes)
    print('\nFirst rows:')
    print(df.head())

inspect_dataset(df)
````
